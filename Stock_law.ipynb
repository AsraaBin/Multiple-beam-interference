```python
from IPython.core.display import HTML

html_layout = r"""
<!DOCTYPE html>
<html lang="en">
<head>
    <script src="https://cdnjs.cloudflare.com/ajax/libs/mathjax/2.7.7/MathJax.js?config=TeX-AMS_CHTML"></script>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <style>
        .simulation-intro {
            font-family: "Times New Roman", Times, serif;
            font-size: 18px;
            line-height: 1.6;
            color: #000000;
            background-color: #ffffff;
            max-width: 900px;
            margin: 10px auto;
            padding: 0 10px;
        }
        .simulation-intro h1, 
        .simulation-intro h2, 
        .simulation-intro h3 {
            font-family: "Times New Roman", Times, serif;
            font-weight: bold;
            margin-top: 1.5em;
            margin-bottom: 0.5em;
        }
        .simulation-intro h1 {
            border-bottom: 2px solid #000000;
            padding-bottom: 10px;
        }
        .simulation-intro h2 {
            border-bottom: 1px solid #000000;
            padding-bottom: 5px;
        }
        .simulation-intro hr {
            border: 0;
            border-top: 1px solid #000000;
            margin: 2em 0;
        }
        .simulation-intro ul {
            margin-top: 0.5em;
            margin-bottom: 1em;
            padding-left: 20px;
        }
        .simulation-intro li {
            margin-bottom: 0.5em;
        }
        .simulation-intro blockquote {
            margin: 1.5em 0;
            padding: 10px 20px;
            border-left: 4px solid #000000;
            font-style: italic;
        }
        .simulation-intro .equation {
            text-align: center;
            margin: 1.5em 0;
        }
    </style>
</head>
<body>
    <div class="simulation-intro">
        <h1>Introduction</h1>
        <p>Multiple-beam interference occurs when light undergoes repeated reflections and transmissions between two parallel interfaces. An incident wave is split into a sequence of coherent beams, each differing in amplitude and phase. The resulting superposition of these beams determines the overall optical response of the system.</p>
        
        <p>This interactive program simulates light propagation inside a dielectric slab. It visualizes in real-time how the slab thickness (\(d\)), refractive index (\(n_2\)), and angle of incidence (\(\theta_i\)) dictate the internal beam paths, optical phase accumulation, and spatial separation between emerging beams.</p>

        <hr>

        <h2>Theoretical Background</h2>

        <h3>1. Refraction and Multiple-Beam Formation</h3>
        <p>At the boundary between two media with refractive indices \(n_1\) and \(n_2\), the direction of the transmitted beam is governed by <strong>Snell’s Law</strong>:</p>
        
        <div class="equation">
            $$n_1 \sin\theta_i = n_2 \sin\theta_t$$
        </div>

        <p>where:</p>
        <ul>
            <li>\(\theta_i\) = Angle of incidence</li>
            <li>\(\theta_t\) = Angle of refraction</li>
        </ul>
        
        <p>Inside a parallel slab, the refracted beam undergoes successive internal reflections. This produces an infinite sequence of emerging parallel beams that share the same angle but possess different optical paths.</p>

        <h3>2. Fresnel Amplitude Splitting</h3>
        <p>At each interface, the incident electric field splits into reflected and transmitted components. The efficiency of this split is determined by the <strong>Fresnel coefficients</strong>:</p>
        
        <div class="equation">
            $$E_r = rE_i \quad \text{and} \quad E_t = tE_i$$
        </div>

        <p>where \(r\) is the reflection coefficient and \(t\) is the transmission coefficient. These depend on the refractive indices, the angle of incidence, and the polarization of the light. For <strong>s-polarized light</strong>, these are expressed as:</p>
        
        <div class="equation">
            $$r_s = \frac{n_1\cos\theta_i - n_2\cos\theta_t}{n_1\cos\theta_i + n_2\cos\theta_t}$$
        </div>
        
        <div class="equation">
            $$t_s = \frac{2n_1\cos\theta_i}{n_1\cos\theta_i + n_2\cos\theta_t}$$
        </div>

        <blockquote>
            <strong>Note:</strong> Because \(|r| < 1\), each successive internal reflection reduces the beam's amplitude, creating a series of progressively weaker coherent waves.
        </blockquote>

        <h3>3. Optical Path Difference and Spatial Shift</h3>
        <p>As beams bounce inside the slab, they accumulate different physical and optical path lengths. The phase difference (\(\delta\)) between two successive emerging beams is given by:</p>
        
        <div class="equation">
            $$\delta = \frac{4\pi n_2 d \cos\theta_t}{\lambda}$$
        </div>

        <p>where:</p>
        <ul>
            <li>\(d\) = Slab thickness</li>
            <li>\(n_2\) = Refractive index of the slab</li>
            <li>\(\lambda\) = Vacuum wavelength</li>
        </ul>

        <p>Alongside phase accumulation, internal reflections cause a lateral spatial shift (\(\Delta x\)) between successive emerging beams:</p>
        
        <div class="equation">
            $$\Delta x = 2d \tan\theta_t$$
        </div>

        <p>Both phase difference and spatial separation are tied to the same geometric and material parameters. You can observe how adjusting these parameters in the simulation alters both beam spacing and interference conditions.</p>

        <h3>4. Reflection Phase &amp; Stokes Relations</h3>
        <p>When a wave reflects off a medium with a <em>higher</em> refractive index (\(n_2 > n_1\)), it undergoes a phase shift of \(\pi\) (equivalent to a path difference of \(\lambda/2\)).</p>
        
        <p>To satisfy energy conservation and time-reversibility in a lossless interface, the forward and reverse scattering amplitudes must obey the <strong>Stokes relations</strong>:</p>
        
        <div class="equation">
            $$r' = -r$$
        </div>
        
        <div class="equation">
            $$tt' = 1 - r^2$$
        </div>

        <p>These constraints ensure that the total energy entering the slab perfectly balances the total energy leaving it through reflection and transmission.</p>
    </div>
</body>
</html>
"""

display(HTML(html_layout))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact
from matplotlib.patches import Arc
from IPython.display import Math, display, HTML
import ipywidgets as widgets

plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = ['DejaVu Serif', 'Liberation Serif', 'Times New Roman']

# Display instructions 
text_content = """
<div style="font-family: 'Times New Roman', Times, serif; font-size: 18px; line-height: 1.5; color: black;">
    <p><strong>Instructions:</strong> Adjust the sliders below to modify the thin-film parameters. 
    Observe how changing the refractive index, slab thickness, and incidence angle 
    alters the internal beam paths, spatial separation, and the resulting reflection/transmission amplitudes.</p>
    
    <br>
    
    <p style="direction: rtl; text-align: right;"><strong>التعليمات:</strong> استخدم أشرطة التمرير أدناه لتعديل خصائص الغشاء الرقيق ديناميكيًا. 
    لاحظ كيف يؤدي تغيير معامل الانكسار (n_film)، وسمك الشريحة (d)، وزاوية السقوط (theta_i) 
    إلى تغيير مسارات الأشعة الداخلية، والتباعد المكاني ، وسعات الانعكاسية والنفاذية الناتجة.</p>
</div>
"""

display(HTML(text_content))

n_film_slider = widgets.FloatSlider(value=1.5, min=1.1, max=2.5, description="n :")
thickness_slider = widgets.IntSlider(value=200, min=50, max=255, description="d :")
theta_inc_slider = widgets.IntSlider(value=25, min=0, max=90, description="θ_i :")

def plot_full_system(n_film=1.5, thickness=200, theta_inc_deg=25):
    n_air = 1.0
    theta_i = np.radians(theta_inc_deg)
    # Snell's Law to find the angle inside the film
    theta_t = np.arcsin((n_air / n_film) * np.sin(theta_i))
    theta_t_deg = np.degrees(theta_t)
    
    fig, ax = plt.subplots(figsize=(12, 6))
    
    # Draw Plate boundaries
    ax.axhline(0, color='black', lw=2, label="Interface 1")
    ax.axhline(-thickness, color='black', lw=2, label="Interface 2")
    
    dx = 2*thickness * np.tan(theta_t)
    x_start = 0

    #Draw the normal 
    ax.axvline(x_start,  color='black', lw=1, label="normal", ls="--" )
    
    #draw the original ray E0
    # Incident ray length
    L_inc = 70
    
    # Incident ray coordinates
    x_inc_start = x_start - L_inc * np.sin(theta_i)
    y_inc_start = L_inc * np.cos(theta_i)
    
    # Draw incident ray
    ax.plot([x_inc_start, x_start], [y_inc_start, 0], 'k-', lw=1.5)
    
    # Label
    ax.text(x_inc_start - 10, y_inc_start + 10, r"$E_0$", fontsize=9)

    #Graphing the angel arc
    # Angle arc radius
    arc_radius = 30
    
    # Draw angle arc
    arc = Arc(
        (x_start, 0),                 # center
        2 * arc_radius,               # width
        2 * arc_radius,               # height
        angle=0,
        theta1=90,                    # start angle
        theta2=90 + theta_inc_deg,    # end angle
        color='#641220',
        lw=1
    )

    # Draw angle arc "Theta t)
    arc_t = Arc(
        (x_start, 0),                 # center
        2 * arc_radius,               # width
        2 * arc_radius,               # height
        angle=0,
        theta1=270,                  # downward normal direction
        theta2=270 + theta_t_deg,    # end angle
        color='black',
        lw=1
    )
    ax.add_patch(arc)
    ax.add_patch(arc_t)

    # Angle label
    ax.text(
        x_start - 15,
        arc_radius + 10,
        r"$\theta_i$",
        fontsize=11,
        color='#641220'
    )

    ax.text(x_start + 10, -20, r"$\theta_t$", fontsize=11, color='black')
    
    for i in range(6):
        # --- Internal Zig-Zags (in the film) ---
        ax.plot([x_start + i*dx, x_start + i*dx + dx/2], [0, -thickness], alpha=1, color='#1f4973')
        ax.plot([x_start + i*dx + dx/2, x_start + i*dx + dx], [-thickness, 0], alpha=1, color='#1f4973') ##0A2A4A  the dark blue 
        
        #adding the Zig Zag label:
        if i>0:
            label_trans = f"$t r'^{i} E_0$"
        else:
            label_trans = f"$t E_0$"
        ax.text((x_start + i*dx + x_start + i*dx + dx/2 + 10)/2, (-thickness+40)/2, label_trans, fontsize=9, rotation=-60, ha='center',va='center')

        ax.text((x_start + i*dx + dx/2 + x_start + i*dx + dx +20)/2,(-thickness)/2, label_trans, fontsize=9, rotation=60,ha='center', va='center')
        
        # --- Reflected Rays  ---
        # Each ray starts slightly shifted 
        ax.plot([x_start + i*dx, x_start + i*dx + 20], [0, 50], lw=1.5, color='#641220')
        # Add labels: t*t'*r^x * E0
        if i > 1:
            label_trans = f"$t t' r'^{2*(i-2)+1} E_0$" 
        elif i == 1: 
            label_trans = "$rE_0$" 
        else:
            label_trans = None
        ax.text(x_start + i*dx -30, 60, label_trans, fontsize=9)
        
        # --- Transmitted Rays (Going down to air) ---
        ax.plot([x_start + i*dx + dx/2, x_start + i*dx + dx/2 + 20], [-thickness, -thickness-50], lw=1.5, color='darkgreen')
        # Add labels: t*t'*r^x * E0
        label_trans = f"$t t' r'^{{{2*i}}} E_0$" if i > 0 else "$t t' E_0$"
        ax.text(x_start + i*dx + dx/2 + 25, -thickness-80, label_trans, fontsize=9)

    #showing t
    x_arrow = -60

    # Double arrow
    ax.annotate(
        '',
        xy=(x_arrow, 0),
        xytext=(x_arrow, -thickness),
        arrowprops=dict(arrowstyle='<->', lw=1.2)
    )
    
    # Thickness label
    ax.text(x_arrow - 15, -thickness/2, r"$t$", fontsize=10)
    
    # Refractive index label
    ax.text(x_arrow + 10, -thickness/2, r"$n_t$", fontsize=10)

    #Calculating the reflection and transmission coefficients 
    # s-Polarization Fresnel Coefficients
    r = (n_air * np.cos(theta_i) - n_film * np.cos(theta_t)) / (n_air * np.cos(theta_i) + n_film * np.cos(theta_t))
    t = (2 * n_air * np.cos(theta_i)) / (n_air * np.cos(theta_i) + n_film * np.cos(theta_t))

    r_prime = (n_film * np.cos(theta_t) - n_air * np.cos(theta_i)) / (n_film * np.cos(theta_t) + n_air * np.cos(theta_i))
    t_prime = (2 * n_film * np.cos(theta_t)) / (n_film * np.cos(theta_t) + n_air * np.cos(theta_i))


    ax.set_ylim(-thickness-150, 150)
    ax.set_xlim(-80, 1200)
    ax.set_title("Interactive Multiple Reflected And Transmitted Beams Tracing", fontsize=16)
    ax.axis('off')
    plt.subplots_adjust(left=0.05, right=0.9, top=0.9, bottom=0.1)
    plt.show()

    # Display the values of the reflection and transmission coefficients
    display(
        HTML(
            f"<div style='text-align: left; font-size: 16px; line-height: 1.6;'>"
            f"The Reflection And Transmission Coefficients (s-polarization):<br>"
            f"<i>r</i> = {r:.3f}<br>"
            f"<i>t</i> = {t:.3f}<br>"
            f"<i>r'</i> = {r_prime:.3f}<br>"
            f"<i>t'</i> = {t_prime:.3f}"
            f"</div>"
         )
     )


# Interactive controls
#interact(plot_full_system, n_film=(1.1, 2.5), thickness=(50, 255), theta_inc_deg=(0, 45));

out = widgets.interactive_output(
    plot_full_system, 
    {'n_film': n_film_slider, 'thickness': thickness_slider, 'theta_inc_deg': theta_inc_slider}
)

display(widgets.VBox([n_film_slider, thickness_slider, theta_inc_slider]), out)
```
